```rust
macro_rules! max {
    ($x: expr) => ($x);
    ($x: expr, $($z: expr),+) => {{
        let y = max!($($z),*);
        if $x > y {
            $x
        } else {
            y
        }
    }}
}

macro_rules! min {
    ($x: expr) => ($x);
    ($x: expr, $($z: expr),+) => {{
        let y = min!($($z),*);
        if $x < y {
            $x
        } else {
            y
        }
    }}
}
```

```rust


fn hex_to_rgb(hex_string: &str) ->  (u8, u8, u8) {
    if hex_string.len() < 7 {
        panic!();
    }
    let h_r = hex_string.get(1..=2).unwrap();
    let h_g = hex_string.get(3..=4).unwrap();
    let h_b = hex_string.get(5..=6).unwrap();
    
    let r = i64::from_str_radix(h_r, 16).unwrap() as u8;
    let g = i64::from_str_radix(h_g, 16).unwrap() as u8;
    let b = i64::from_str_radix(h_b, 16).unwrap() as u8;
    
    (r,g,b)
}

fn hex_to_hsl(hex_string: &str) -> (u8, u8, u8) {
    let (r, g, b) = hex_to_rgb(hex_string);
    let normalized_r = r as f32 / 255.0;
    let normalized_g = g as f32 / 255.0;
    let normalized_b = b as f32 / 255.0;

    let min = min!(normalized_r, normalized_g, normalized_b);
    let max = max!(normalized_r, normalized_g, normalized_b);

    
    let l = (min + max) / 2.0;

    let s = if min == max {
        0.0
    } else if l <= 0.5 {
        (max - min) / (max + min)
    } else {
        (max - min) / (2.0 - max - min)
    };

    /*
    If Red is max, then Hue = (G-B)/(max-min)
    If Green is max, then Hue = 2.0 + (B-R)/(max-min)
    If Blue is max, then Hue = 4.0 + (R-G)/(max-min)
    */
    let mut h = match max {
        (normalized_r) => (normalized_g - normalized_b) / (max - min),
        (normalized_g) => 2.0 + (normalized_b - normalized_r) / (max - min),
        (normalized_b) => 4.0 + (normalized_r - normalized_g) / (max - min),
    };

    h *= 60.0;

    if(h < 0.0) {
        h += 360.0;
    }
    
    ((h * 100.0) as u8, (s * 100.0) as u8, (l * 100.0) as u8)   
}
```

```rust
#[derive(Debug)]
enum ColorMode {
    Hex(String),
    Rgb((u8, u8, u8)),
    Hsl((u8, u8, u8)),
}

#[derive(Debug)]
struct Color {
    hex: ColorMode,
    rgb: ColorMode,
    hsl: ColorMode,
}

impl Color {
    fn from_hex(hex_string: &str) -> Self {
        let rgb = hex_to_rgb(hex_string);
        let hsl = hex_to_hsl(hex_string);
        let hex = hex_string.into();
        
        Color { 
            hex : ColorMode::Hex(hex), 
            rgb : ColorMode::Rgb(rgb), 
            hsl : ColorMode::Hsl(hsl), 
        }
    }
}
```

```rust
let color_a = Color::from_hex("#ff0000");
color_a.rgb
```
